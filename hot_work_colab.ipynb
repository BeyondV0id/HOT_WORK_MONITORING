```python
# Step 4: Read and Display Incident Log CSV
import os
import pandas as pd

csv_path = 'logs/incidents.csv'
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    display(df)
else:
    print('No CSV incident log file found yet.')
```

```python
# Step 5: Render Violation Snapshots Directly in Notebook
import glob
import cv2
import matplotlib.pyplot as plt

snapshot_files = glob.glob('logs/snapshots/*.jpg')

if snapshot_files:
    print(f'Found {len(snapshot_files)} violation snapshot(s):\n')
    for img_path in snapshot_files[:5]:
        img = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        plt.figure(figsize=(10, 6))
        plt.imshow(img_rgb)
        plt.title(f'Violation Snapshot: {os.path.basename(img_path)}')
        plt.axis('off')
        plt.show()
else:
    print('No violation snapshots found in logs/snapshots/')
```
